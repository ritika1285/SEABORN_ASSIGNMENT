```python
### SESSION 6 - Jointplot ###

#Task 1

#Question:Load a CSV file containing Flipkart product ratings and prices, then use Seaborn's jointplot with kind='scatter' to visualize the relationship between rating and price.

#Answer:

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load the CSV file
df = pd.read_csv("flipkart_products.csv")
print(df.head())

print("="*100)

# Create scatter jointplot
sns.jointplot(
    x='Rating',
    y='actprice1',
    data=df,
    kind='scatter'
)

plt.show()
```

```python
##Task 2

#Question:Create a jointplot with kind='hex' to analyze the relationship between delivery time and order value using a dataset of Swiggy food orders.

#Answer:

# Sample Swiggy order data
data = {
    'delivery_time': [25, 30, 35, 40, 28, 32, 45, 38, 27, 31,
                      36, 42, 29, 33, 37, 41, 26, 34, 39, 30],
    'order_value': [250, 300, 450, 500, 280, 350, 600, 550, 220, 320,
                    400, 520, 270, 380, 430, 480, 200, 360, 470, 310]
}

df = pd.DataFrame(data)

# Create hexbin jointplot
sns.jointplot(
    x='delivery_time',
    y='order_value',
    data=df,
    kind='hex'
)

plt.show()
```

```python
##Task 4

#Question:Compare the visual insights you get from scatter, hexbin, and KDE jointplots using the same dataset of Zomato restaurant reviews (rating vs number of reviews). Briefly describe which plot type helps you understand the data best and why.

#Answer:

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Sample Zomato restaurant review data
data = {
    'rating': [3.2, 3.5, 3.8, 4.0, 4.1, 4.2, 4.3, 4.5, 4.6, 4.7,
               3.4, 3.7, 3.9, 4.1, 4.2, 4.4, 4.5, 4.6, 4.8, 4.9],
    'reviews': [120, 250, 400, 550, 700, 850, 1000, 1200, 1500, 1800,
                180, 320, 450, 650, 800, 1100, 1350, 1600, 1900, 2200]
}

df = pd.DataFrame(data)

#1. Scatter Jointplot
sns.jointplot(
    x='rating',
    y='reviews',
    data=df,
    kind='scatter'
)
plt.show()
```

```python
#2. Hexbin Jointplot
sns.jointplot(
    x='rating',
    y='reviews',
    data=df,
    kind='hex'
)

plt.show()
```

```python
#3. KDE Jointplot
sns.jointplot(
    x='rating',
    y='reviews',
    data=df,
    kind='kde'
)

plt.show()
```

### Comparison

| Jointplot Type | What It Shows |
|---|---|
| `scatter` | Individual data points and their relationship |
| `hex` | Concentration of data points using hexagonal areas |
| `kde` | Smooth density and concentration of observations |


##Observation
The scatter plot is useful for seeing individual observations and the overall relationship between rating and number of reviews.

The hexbin plot is useful when many points overlap because it shows areas where observations are concentrated.

The KDE plot provides a smooth view of data density and helps identify where most observations are concentrated.

##Conclusion
For this dataset, the scatter plot provides the most direct view of the relationship because the dataset is relatively small and individual observations can be seen clearly. For a larger dataset with many overlapping points, a hexbin or KDE plot can provide a clearer view of the data concentration.
