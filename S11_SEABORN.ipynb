```python
#Task 1

#Question:Create a Seaborn barplot showing the number of orders per food category (e.g., Pizza, Burger, Biryani, Momos, Ice Cream) and save the chart as a PNG file with 300 DPI.

#Answer:

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.DataFrame({
    'category': ['Pizza', 'Burger', 'Biryani', 'Momos', 'Ice Cream'],
    'orders': [250, 300, 220, 180, 150]
})

plt.figure(figsize=(8, 5))

sns.barplot(
    x='category',
    y='orders',
    data=df
)

plt.xlabel("Food Category")
plt.ylabel("Number of Orders")
plt.title("Orders by Food Category")

plt.savefig("food_orders_300dpi.png", dpi=300, bbox_inches='tight')

plt.show()
```

```python
#Task 2

#Question:Generate a Seaborn line plot visualizing daily step counts for one week, then export the chart as an SVG file suitable for use in a PowerPoint presentation.
#Answer:

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.DataFrame({
    'day': ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun'],
    'steps': [4500, 5200, 6000, 5500, 7000, 8500, 7500]
})

plt.figure(figsize=(9, 5))

sns.lineplot(
    x='day',
    y='steps',
    data=df,
    marker='o'
)

plt.xlabel("Day")
plt.ylabel("Steps")
plt.title("Daily Step Count")

plt.savefig("weekly_steps.svg", format='svg', bbox_inches='tight')

plt.show()
```

```python
#Task 3

#Question:Customize a Seaborn chart using the 'darkgrid' style and a color palette of your choice, then save it as a high-resolution PNG image with 1200 DPI.

#Answer:

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.DataFrame({
    'month': ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun'],
    'revenue': [45000, 52000, 58000, 61000, 68000, 75000]
})

sns.set_style("darkgrid")
sns.set_palette("deep")

plt.figure(figsize=(10, 6))

sns.barplot(
    x='month',
    y='revenue',
    data=df
)

plt.xlabel("Month")
plt.ylabel("Revenue")
plt.title("Monthly Revenue")

plt.savefig(
    "monthly_revenue_1200dpi.png",
    dpi=1200,
    bbox_inches='tight'
)

plt.show()
```

```python
#Task 4

#Question:Create a template function called apply_presentation_theme() that applies a consistent Seaborn style, font, and color palette to any chart, then use it to style and export a pie chart showing your top 5 most-used apps.
#Answer:

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

def apply_presentation_theme():
    sns.set(
        context='talk',
        style='whitegrid',
        palette='deep',
        font_scale=1.0
    )

# Apply presentation theme
apply_presentation_theme()

# Top 5 most-used apps
apps = ['Instagram', 'YouTube', 'WhatsApp', 'Spotify', 'Google']
usage = [30, 25, 20, 15, 10]

plt.figure(figsize=(5, 5))

plt.pie(
    usage,
    labels=apps,
    autopct='%1.1f%%',
    startangle=90
)

plt.title("Top 5 Most-Used Apps")

plt.savefig(
    "top_5_apps.png",
    dpi=300,
    bbox_inches='tight'
)

plt.show()
```

```python
#Task 5

#Question:Use ChatGPT to generate a code snippet for exporting a Seaborn heatmap as a 1200 DPI SVG file, then test and adapt the code to save your own chart.

#Answer:

# ChatGPT Prompt:
# "Generate Python code to create a Seaborn heatmap
# and save it as a high-quality SVG file."

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.DataFrame({
    'Sales': [100, 200, 300, 400],
    'Quantity': [10, 20, 30, 40],
    'Profit': [20, 40, 60, 80],
    'Discount': [5, 10, 15, 20]
})

correlation_matrix = df.corr()

plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title("Sales Correlation Heatmap")

plt.savefig(
    "sales_heatmap.svg",
    format='svg',
    dpi=1200,
    bbox_inches='tight'
)

plt.show()
```
