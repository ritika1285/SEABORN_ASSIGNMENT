```python
#### SESSION : 4

#Task 1

#Question:Load a CSV file containing Flipkart product prices and use sns.boxplot() to visualize the price distribution.

#Answer:

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load the CSV file
df = pd.read_csv("flipkart_products.csv")

# Display the first few rows
print(df.head())

# Create boxplot
sns.boxplot(y='price', data=df)

plt.xlabel("Product Price")
plt.ylabel("Price (₹)")
plt.title("Distribution of Flipkart Product Prices")

plt.show()
```

```python
#Task 2

#Question:Use sns.violinplot() to compare the order amounts for two different Zomato restaurant categories (e.g., Fast Food vs Desserts) from a sample dataset.

#Answer:

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    'restaurant_category': [
        'Fast Food', 'Fast Food', 'Fast Food', 'Fast Food', 'Fast Food',
        'Desserts', 'Desserts', 'Desserts', 'Desserts', 'Desserts'
    ],
    'order_amount': [
        250, 300, 350, 400, 450,
        150, 200, 250, 300, 350
    ]
}

df = pd.DataFrame(data)

sns.violinplot(
    x='restaurant_category',
    y='order_amount',
    data=df
)

plt.xlabel("Restaurant Category")
plt.ylabel("Order Amount (₹)")
plt.title("Order Amount Distribution by Restaurant Category")

plt.show()
```

```python
##Task 3

#Question:Given an array of IPL match ticket prices, plot a boxplot using sns.boxplot() and identify any outliers in the data.

#Answer:

ticket_prices = [
    500, 600, 700, 800, 900,
    1000, 1100, 1200, 1300, 1400,
    1500, 1600, 1700, 1800, 2000,
    2200, 2500, 3000, 8000
]

sns.boxplot(y=ticket_prices)

plt.ylabel("Ticket Price (₹)")
plt.title("IPL Match Ticket Price Distribution")

plt.show()
```

Observation:

The value ₹8,000 appears as a point outside the upper whisker of the boxplot. Therefore, it can be identified as a potential outlier.

```python
##Task 4

#Question:Create a violin plot using sns.violinplot() to show the distribution of daily step counts collected from a fitness app like Google Fit, and explain what the shape of the plot tells you about user activity.

#Answer:

step_counts = [
    3500, 4200, 5000, 5500, 6000,
    6200, 6500, 6800, 7000, 7200,
    7500, 7800, 8000, 8200, 8500,
    9000, 9500, 10000, 11000, 12000,
    4500, 5300, 6100, 6700, 7300,
    7600, 8100, 8800, 9200, 10500
]

sns.violinplot(y=step_counts)

plt.ylabel("Daily Steps")
plt.title("Distribution of Daily Step Counts")

plt.show()
```

Observation:

The wider parts of the violin represent step-count values that occur more frequently, while narrower parts represent less frequent values.

The shape helps us understand the distribution of user activity, such as whether most users have a similar number of daily steps or whether the activity levels are spread across a wider range.

Key Difference

Plot	                 Main Purpose
sns.boxplot()	         Shows median, quartiles, spread, and potential outliers
sns.violinplot()	     Shows the distribution and density of the data
