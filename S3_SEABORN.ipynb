```python
### SESSION 3 - Distribution Plots ###

##Task 1

#Question: Import Seaborn and use sns.histplot() to visualize the distribution of delivery times (in minutes) from a sample Swiggy order dataset with at least 30 values.

#Answer:

import seaborn as sns
import matplotlib.pyplot as plt

delivery_times = [
    25, 30, 35, 28, 40, 32, 45, 38, 27, 31,
    36, 42, 29, 33, 37, 41, 26, 34, 39, 30,
    35, 43, 28, 32, 36, 40, 29, 31, 38, 44
]

sns.histplot(delivery_times, bins=10, kde=False)

plt.xlabel("Delivery Time (minutes)")
plt.ylabel("Number of Orders")
plt.title("Distribution of Delivery Times")

plt.show()
```

```python
##Task 2

#Question: Download or create a list of daily step counts (e.g., from a fitness app) for one month and plot the distribution using sns.kdeplot(), showing the probability density curve.

#Answer:

step_counts = [
    4500, 5200, 6100, 7000, 5500, 8000, 7500,
    6200, 6800, 7200, 5000, 5800, 6500, 9000,
    8500, 7300, 6900, 5600, 6400, 7800, 8200,
    4700, 5900, 6700, 7100, 7600, 8800, 5300,
    6000, 7400
]

sns.kdeplot(step_counts, fill=True)

plt.xlabel("Daily Steps")
plt.ylabel("Probability Density")
plt.title("Distribution of Daily Step Counts")

plt.show()
```

```python
## Task 3

#Question: Combine sns.histplot() and sns.kdeplot() to visualize the distribution of monthly spending amounts (in rupees) on Zomato orders, using the same dataset for both plots on a single figure.

#Answer:

monthly_spending = [
    1200, 1500, 1800, 2200, 2500,
    1300, 1700, 2000, 2800, 3200,
    1900, 2300, 2600, 3000, 3500,
    1400, 1600, 2100, 2400, 2900,
    1100, 1750, 2250, 2700, 3100,
    1350, 1850, 2150, 2550, 3300
]

sns.histplot(monthly_spending, bins=10, kde=False)

sns.kdeplot(monthly_spending, fill=False, linewidth=2)

plt.xlabel("Monthly Spending (₹)")
plt.ylabel("Frequency / Density")
plt.title("Distribution of Monthly Zomato Spending")

plt.show()
```

```python
## Task 4

#Question: Given a dataset of IPL match scores, use sns.histplot() to plot the run distribution. Change the number of bins and observe how the plot changes. Write 2–3 lines about what you notice.

#Answer:

#Using bins=5

ipl_scores = [
    120, 145, 156, 178, 190, 135, 160, 172, 185, 150,
    142, 165, 180, 195, 155, 130, 170, 182, 148, 158,
    175, 188, 200, 140, 162, 177, 192, 150, 168, 185
]

sns.histplot(ipl_scores, bins=5)

plt.xlabel("Runs")
plt.ylabel("Number of Matches")
plt.title("IPL Run Distribution - 5 Bins")

plt.show()

print("==================================================================")
print("==================================================================")

#Using bins=10
sns.histplot(ipl_scores, bins=10)

plt.xlabel("Runs")
plt.ylabel("Number of Matches")
plt.title("IPL Run Distribution - 10 Bins")

plt.show()

print("==================================================================")
print("==================================================================")
#Using bins=20
sns.histplot(ipl_scores, bins=20)

plt.xlabel("Runs")
plt.ylabel("Number of Matches")
plt.title("IPL Run Distribution - 20 Bins")

plt.show()
```

## Observation ##
When the number of bins is increased, the histogram shows more detailed information about the distribution of runs.

With fewer bins, the distribution looks simpler and more general, while more bins show smaller differences between the run values.
##Conclusion

bins controls how many intervals are used to divide the data. A smaller number of bins gives a simpler view, while a larger number of bins provides more detail about the distribution.
