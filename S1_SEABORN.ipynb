```python
import seaborn as sns
import matplotlib.pyplot as plt

print("Seaborn version:", sns.__version__)
print("Matplotlib version:", plt.matplotlib.__version__)
```

```python
#Task 2 — Weekly Temperature Trend Using sns.lineplot()

days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
temps = [32, 34, 33, 35, 36, 37, 34]

sns.lineplot(x=days, y=temps, marker='o')

plt.xlabel("Day")
plt.ylabel("Temperature (°C)")
plt.title("Weekly Temperature Trend")

plt.show()
```

```python
#Task 3 — Recreate the above line plot using Seaborn's axes-level and figure-level functions separately. Briefly note one difference you observe in how these functions work.
#A. Axes-Level Function — sns.lineplot()

days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
temps = [32, 34, 33, 35, 36, 37, 34]

sns.lineplot(x=days, y=temps, marker='o')

plt.xlabel("Day")
plt.ylabel("Temperature (°C)")
plt.title("Weekly Temperature Trend")

plt.show()

print("========================================================================")
print("========================================================================")

#B. Figure-Level Function — sns.relplot()

days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
temps = [32, 34, 33, 35, 36, 37, 34]

sns.relplot(
    x=days,
    y=temps,
    kind='line',
    marker='o'
)

plt.xlabel("Day")
plt.ylabel("Temperature (°C)")
plt.title("Weekly Temperature Trend")

plt.show()
```

Difference

Axes-level: sns.lineplot() draws the plot on a specific Matplotlib axes and gives more direct control over an individual plot.

Figure-level: sns.relplot() creates and manages the complete figure and is useful for creating plots across multiple subsets using options such as col and row.

```python
#Task 4 Apply each of Seaborn's built-in themes (darkgrid, whitegrid, ticks) to your temperature trend plot and save three separate images. Compare how the grid and background change with each theme

#1. Darkgrid Theme

days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
temps = [32, 34, 33, 35, 36, 37, 34]

sns.set_theme(style='darkgrid')

plt.figure()
sns.lineplot(x=days, y=temps, marker='o')

plt.xlabel("Day")
plt.ylabel("Temperature (°C)")
plt.title("Weekly Temperature Trend - Darkgrid")

plt.savefig("temperature_darkgrid.png")
plt.show()

print("========================================================================")
print("========================================================================")

#2. Whitegrid Theme

sns.set_theme(style='whitegrid')

plt.figure()
sns.lineplot(x=days, y=temps, marker='o')

plt.xlabel("Day")
plt.ylabel("Temperature (°C)")
plt.title("Weekly Temperature Trend - Whitegrid")

plt.savefig("temperature_whitegrid.png")
plt.show()

print("========================================================================")
print("========================================================================")

#3. Ticks Theme
sns.set_theme(style='ticks')

plt.figure()
sns.lineplot(x=days, y=temps, marker='o')

plt.xlabel("Day")
plt.ylabel("Temperature (°C)")
plt.title("Weekly Temperature Trend - Ticks")

plt.savefig("temperature_ticks.png")
plt.show()
```

### Theme Comparison

| Theme | Background | Grid |
|---|---|---|
| darkgrid | Gray | Visible grid |
| whitegrid | White | Visible grid |
| ticks | White | No major grid; axis ticks are emphasized |

### Conclusion

Seaborn themes change the overall appearance of a chart, especially the background, grid lines, and axis ticks, without changing the underlying data.
