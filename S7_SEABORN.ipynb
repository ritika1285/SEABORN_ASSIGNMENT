```python
### SESSION 7 - Heatmaps & Correlation Analysis ###

##Task 1

#Question:Create a Pandas DataFrame with at least 5 columns representing different features of food delivery orders (e.g., delivery_time, order_value, tip_amount, rating, delivery_distance) and 20 rows of random data, then calculate and display the correlation matrix using the .corr() function.

#Answer:

import pandas as pd
import numpy as np

np.random.seed(42)

df = pd.DataFrame({
    'delivery_time': np.random.randint(15, 60, 20),
    'order_value': np.random.randint(200, 1500, 20),
    'tip_amount': np.random.randint(20, 200, 20),
    'rating': np.round(np.random.uniform(3.0, 5.0, 20), 1),
    'delivery_distance': np.round(np.random.uniform(1, 15, 20), 2)
})

print("Food Delivery Data:")
display(df)
print("="*50)

correlation_matrix = df.corr()

print("Correlation Matrix:")
display(correlation_matrix)
```

```python
##Task 2

#Question:Using Seaborn, plot a heatmap of the correlation matrix you generated for the food delivery data, and add annotations to show the correlation values inside each cell.

#Answer:

import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title("Food Delivery Correlation Heatmap")
plt.show()
```

```python
##Task 3

#Question:Change the color palette of your Seaborn heatmap to a diverging palette (such as 'coolwarm' or 'RdYlGn') to better visualize positive and negative correlations, and explain which two features in your data have the strongest negative correlation.
#Hint: Use the cmap parameter in sns.heatmap().

#Answer:

plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0
)

plt.title("Food Delivery Correlation Heatmap")
plt.show()
```

```python
#To find the strongest negative correlation:

corr = correlation_matrix.copy()

# Remove diagonal values
np.fill_diagonal(corr.values, np.nan)

# Find the strongest negative correlation
strongest_negative = corr.stack().idxmin()
negative_value = corr.stack().min()

print("Strongest Negative Correlation:")
print(strongest_negative[0], "and", strongest_negative[1])
print("Correlation:", round(negative_value, 2))

#Observation:The two features displayed by the code have the strongest negative correlation in the generated dataset. A negative correlation means that when one feature increases, the other tends to decrease.
```

```python
##Task 4

#Question:Take the Flipkart mobile phones dataset (or any small CSV with product features you can find online), compute the correlation matrix for at least 4 numeric columns (e.g., price, rating, RAM, battery_capacity), and generate a Seaborn heatmap with annotations and a diverging color palette.

#Answer:

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load the CSV file
df_mobile = pd.read_csv("flipkart_mobile_phones.csv")

# Select at least 4 numeric columns
numeric_columns = [
    'price',
    'rating',
    'ram',
    'battery_capacity'
]

# Calculate correlation matrix
correlation_matrix = df_mobile[numeric_columns].corr()

print("Correlation Matrix:")
display(correlation_matrix)

# Create heatmap
plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0
)

plt.title("Flipkart Mobile Phones Correlation Heatmap")
plt.show()
```

#Observation:
The heatmap shows the relationships between price, rating, RAM, and battery capacity. Positive correlation indicates that two features tend to increase together, while negative correlation indicates an inverse relationship.

#Conclusion

Correlation analysis helps identify relationships between numerical variables. A heatmap makes these relationships easy to understand visually, and a diverging color palette helps distinguish positive and negative correlations.
