```python
##Task 1

#Question:Use sns.scatterplot() to visualize the relationship between the number of followers and the number of posts for 20 fictional Instagram users. Label the axes appropriately.

#Answer:

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

np.random.seed(42)

df = pd.DataFrame({
    'followers': np.random.randint(500, 10000, 20),
    'posts': np.random.randint(10, 500, 20)
})

plt.figure(figsize=(8, 5))

sns.scatterplot(
    x='followers',
    y='posts',
    data=df
)

plt.xlabel("Number of Followers")
plt.ylabel("Number of Posts")
plt.title("Followers vs Number of Posts")

plt.show()
```

```python
#Task 2

#Question:Create a line plot using sns.lineplot() to show the daily number of food orders on Zomato for each day over a month. Use a list of 30 random order counts to simulate the data.

##Answer:

import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

np.random.seed(42)

days = list(range(1, 31))
orders = np.random.randint(50, 200, 30)

plt.figure(figsize=(10, 5))

sns.lineplot(
    x=days,
    y=orders,
    marker='o'
)

plt.xlabel("Day of the Month")
plt.ylabel("Number of Orders")
plt.title("Daily Zomato Orders Over a Month")

plt.show()
```

```python
#Task 3

#Question:Using sns.relplot(), plot the relationship between movie ratings and the number of reviews for different genres (Action, Comedy, Drama) based on a sample dataset of 15 movies. Use the hue parameter to differentiate genres.
#Answer:

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.DataFrame({
    'movie': [
        'Movie 1', 'Movie 2', 'Movie 3', 'Movie 4', 'Movie 5',
        'Movie 6', 'Movie 7', 'Movie 8', 'Movie 9', 'Movie 10',
        'Movie 11', 'Movie 12', 'Movie 13', 'Movie 14', 'Movie 15'
    ],
    'rating': [
        4.5, 4.2, 3.8, 4.7, 3.9,
        4.1, 4.6, 3.7, 4.3, 4.8,
        3.6, 4.4, 4.0, 3.9, 4.5
    ],
    'reviews': [
        1200, 950, 700, 1500, 800,
        1100, 1350, 600, 1000, 1800,
        500, 1250, 850, 750, 1400
    ],
    'genre': [
        'Action', 'Action', 'Action', 'Action', 'Action',
        'Comedy', 'Comedy', 'Comedy', 'Comedy', 'Comedy',
        'Drama', 'Drama', 'Drama', 'Drama', 'Drama'
    ]
})

sns.relplot(
    x='rating',
    y='reviews',
    hue='genre',
    data=df,
    kind='scatter',
    height=5,
    aspect=1.5
)

plt.xlabel("Movie Rating")
plt.ylabel("Number of Reviews")
plt.title("Movie Ratings vs Number of Reviews")

plt.show()
```

```python
##Task 4

#Question:Generate a regression plot using sns.regplot() to analyze the correlation between the number of hours spent on YouTube and exam scores for a group of 25 students. Interpret whether the trend is positive or negative.

#Answer:

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

np.random.seed(42)

hours = np.random.uniform(1, 10, 25)

exam_scores = 45 + (hours * 5) + np.random.normal(0, 5, 25)

df = pd.DataFrame({
    'youtube_hours': hours,
    'exam_score': exam_scores
})

plt.figure(figsize=(8, 5))

sns.regplot(
    x='youtube_hours',
    y='exam_score',
    data=df
)

plt.xlabel("Hours Spent on YouTube")
plt.ylabel("Exam Score")
plt.title("YouTube Hours vs Exam Scores")

plt.show()
```

#Observation:
The regression line shows a positive trend in this simulated dataset. As YouTube hours increase, the exam scores tend to increase. This is only a simulated relationship and does not establish that YouTube use causes higher exam scores.

#Conclusion

Relational plots help analyze relationships between variables. scatterplot() shows individual observations, lineplot() shows trends over time, relplot() can compare relationships across categories, and regplot() adds a regression line to show the overall linear trend.
