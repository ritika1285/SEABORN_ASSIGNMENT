```python
#Task 1

#Question:Download the Kaggle Bank Marketing Dataset and use Pandas to load it into a DataFrame named bank_df, then display the first 10 rows and the DataFrame's info().

#Answer:

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# The dataset
bank_df = pd.read_excel("C:\\Users\\Admin\\Downloads\\BANK.xlsx")

# Display first 10 rows
print(bank_df.head(10))

# Display DataFrame information
bank_df.info()
```

```python
#Task 2

#Question:Use Seaborn's countplot to visualize how many customers in the dataset subscribed to a term deposit (column y), and display the plot with appropriate axis labels.

#Answer:

plt.figure(figsize=(7, 5))

sns.countplot(
    x='y',
    data=bank_df
)

plt.xlabel("Term Deposit Subscription")
plt.ylabel("Number of Customers")
plt.title("Term Deposit Subscription Distribution")

plt.show()
```

```python
#Task 3

#Question:Analyze the relationship between customer age and term deposit subscription by creating a Seaborn boxplot with y on the x-axis and age on the y-axis.

#Answer:

plt.figure(figsize=(8, 5))

sns.boxplot(
    x='y',
    y='age',
    data=bank_df
)

plt.xlabel("Term Deposit Subscription")
plt.ylabel("Age")
plt.title("Age vs Term Deposit Subscription")

plt.show()
```

```python
#Task 4

#Question:Create a bar plot using Seaborn to show the percentage of customers who subscribed to a term deposit for each education level (column education).

#Answer:

# Calculate subscription percentage for each education level
education_subscription = (
    bank_df.groupby('education')['y']
    .apply(lambda x: (x == 'yes').mean() * 100)
    .reset_index(name='subscription_percentage')
)

print(education_subscription)
print("="*50)
plt.figure(figsize=(9, 5))

sns.barplot(
    x='education',
    y='subscription_percentage',
    data=education_subscription
)

plt.xlabel("Education Level")
plt.ylabel("Subscription Percentage (%)")
plt.title("Term Deposit Subscription Percentage by Education")

plt.xticks(rotation=20)
plt.show()
```

```python
#Task 5

#Question:Use ChatGPT or Copilot to suggest two additional visualizations that could help a fintech startup like Cred understand which users are most likely to buy a new savings product, then implement one of those visualizations using Seaborn.

#Answer:

#Two additional visualization ideas:
#Job vs Subscription Rate
#Shows which job categories have higher term-deposit subscription percentages.
#Useful for identifying customer segments.
#Balance vs Subscription
#A boxplot comparing account balance for customers who subscribed and did not subscribe.
#Useful for understanding the relationship between account balance and subscription.
#Implementation: Job vs Subscription Rate
# Calculate subscription percentage by job
job_subscription = (
    bank_df.groupby('job')['y']
    .apply(lambda x: (x == 'yes').mean() * 100)
    .reset_index(name='subscription_percentage')
)

print(job_subscription)
print("="*100)

plt.figure(figsize=(12, 6))

sns.barplot(
    x='job',
    y='subscription_percentage',
    data=job_subscription
)

plt.xlabel("Job")
plt.ylabel("Subscription Percentage (%)")
plt.title("Term Deposit Subscription Rate by Job")

plt.xticks(rotation=45)
plt.show()
```
