```python
### SESSION 2 - Categorical Visuals (Countplot, Barplot) ###

#Task 1
#Question:Use sns.countplot() to visualize the number of orders placed by each payment method (e.g., UPI, Card, Cash) in a sample Swiggy orders DataFrame with at least 3 payment types.

#Answer:

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    'payment_method': ['UPI', 'Card', 'Cash', 'UPI', 'Card',
                       'UPI', 'Cash', 'Card', 'UPI', 'Cash']
}

df = pd.DataFrame(data)

sns.countplot(x='payment_method', data=df)

plt.xlabel('Payment Method')
plt.ylabel('Number of Orders')
plt.title('Number of Orders by Payment Method')

plt.show()
```

```python
#Task 2

#Question:Create a barplot using sns.barplot() to show the average delivery time (in minutes) for each food category (e.g., Pizza, Burger, Biryani, South Indian) from a Zomato-style dataset.

#Answer:

data = {
    'food_category': ['Pizza', 'Burger', 'Biryani', 'South Indian',
                      'Pizza', 'Burger', 'Biryani', 'South Indian'],
    'delivery_time': [35, 30, 40, 25, 45, 28, 38, 30]
}

df = pd.DataFrame(data)

sns.barplot(
    x='food_category',
    y='delivery_time',
    data=df,
    estimator='mean'
)

plt.xlabel('Food Category')
plt.ylabel('Average Delivery Time (minutes)')
plt.title('Average Delivery Time by Food Category')

plt.show()
```

```python
#Task 3

#Question: Modify your barplot to display the total number of orders instead of the average by setting estimator to sum in sns.barplot().

#Answer:

data = {
    'food_category': ['Pizza', 'Burger', 'Biryani', 'South Indian'],
    'orders': [120, 150, 100, 80]
}

df = pd.DataFrame(data)

sns.barplot(
    x='food_category',
    y='orders',
    data=df,
    estimator='sum'
)

plt.xlabel('Food Category')
plt.ylabel('Total Number of Orders')
plt.title('Total Orders by Food Category')

plt.show()
```

```python
#Task 4

#Question: Plot a countplot of movie genres from a BookMyShow-style dataset, but remove the confidence interval from the plot.

#Answer:

data = {
    'genre': ['Action', 'Comedy', 'Drama', 'Action',
              'Romance', 'Comedy', 'Action', 'Drama',
              'Comedy', 'Romance']
}

df = pd.DataFrame(data)

sns.countplot(x='genre', data=df)

plt.xlabel('Movie Genre')
plt.ylabel('Number of Movies')
plt.title('Number of Movies by Genre')

plt.show()
```
