```python
#Task 1

#Question:Use sns.set() to apply the 'poster' context to a Seaborn barplot showing the number of songs streamed per day on Spotify, and observe how the plot elements change compared to the default context.

#Answer:

import seaborn as sns
import matplotlib.pyplot as plt

days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
streams = [120, 150, 180, 160, 200, 250, 220]

sns.set(context='poster')

plt.figure(figsize=(10, 6))

sns.barplot(
    x=days,
    y=streams
)

plt.xlabel("Day")
plt.ylabel("Songs Streamed")
plt.title("Daily Spotify Songs Streamed")

plt.show()
```

```python
#Task 2

#Question:Create a Seaborn countplot showing the distribution of food orders by cuisine type (e.g., Indian, Chinese, Italian) using a 'colorblind' palette, similar to how Zomato might visualize order trends.
#Answer:

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.DataFrame({
    'cuisine': [
        'Indian', 'Chinese', 'Italian', 'Indian', 'Chinese',
        'Indian', 'Italian', 'Chinese', 'Indian', 'Italian',
        'Chinese', 'Indian', 'Italian', 'Indian', 'Chinese',
        'Italian', 'Indian', 'Chinese', 'Indian', 'Italian'
    ]
})

plt.figure(figsize=(8, 5))

sns.countplot(
    x='cuisine',
    data=df,
    palette='colorblind'
)

plt.xlabel("Cuisine Type")
plt.ylabel("Number of Orders")
plt.title("Food Orders by Cuisine Type")

plt.show()
```

```python
#Task 4

#Question:Refactor an existing Seaborn plot by switching its palette from 'deep' to 'colorblind' and changing the context from 'talk' to 'poster'. Briefly describe the visual differences you notice after applying these changes.

#Answer:

import seaborn as sns
import matplotlib.pyplot as plt

days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
sales = [120, 150, 180, 140, 200, 230, 210]

# Updated styling
sns.set(
    context='poster',
    palette='colorblind'
)

plt.figure(figsize=(10, 6))

sns.barplot(
    x=days,
    y=sales
)

plt.xlabel("Day")
plt.ylabel("Sales")
plt.title("Daily Sales")

plt.show()
```

```python
#Task 5

#Question:Design a Seaborn theme that mimics the look of a Flipkart sales dashboard: use sns.set() to set a 'deep' palette, 'talk' context, and customize the font scale and grid style. Apply your theme to a barplot showing sales of 5 product categories.

#Answer:

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.DataFrame({
    'category': [
        'Mobiles',
        'Electronics',
        'Fashion',
        'Home Appliances',
        'Grocery'
    ],
    'sales': [
        850000,
        720000,
        560000,
        640000,
        480000
    ]
})

# Create corporate-style Seaborn theme
sns.set(
    context='talk',
    style='whitegrid',
    palette='deep',
    font_scale=1.0
)

plt.figure(figsize=(10, 6))

sns.barplot(
    x='category',
    y='sales',
    data=df
)

plt.xlabel("Product Category")
plt.ylabel("Sales (₹)")
plt.title("Flipkart Sales Dashboard")

plt.xticks(rotation=15)
plt.show()
```
